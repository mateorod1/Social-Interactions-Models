### Introducción a recorridos

Los recorridos se realizan sobre iterables que conocemos o sobre condiciones lógicas que establecemos desde un principio. Cabe aclarar que los recorridos son la capa más externa de lo que podemos entender como un algoritmo, por lo cuál se debe asegurar que el recorrido sea finito. Por ahora tendremos dos tipos de recorrido relevante: for y while.

#### Recorridos tipo: for

```python
# Recorrido tipo: for
Iterable_A_Recorrer = [1,2,3]

for indice in Iterable_A_Recorrer:
    print("Se realizan acciones sobre 'algo' haciendo uso del indice: {}".format(indice))
    print(f"Indice: {indice}")
```

```python
# Probemos distintos recorridos sobre distintos tipos de iterables
Iterable1 = [1,2,3,"Hola","Mundo"]
Iterable2 = (1,2,3,4)
Iterable3 = {"llave1":"hola","llave2":"mundo","llave3":[1,2,3]}
```

```python
# Recorridos Simples (1)
for i in Iterable1:
    print(i)
```

```python
# Recorridos Simples (2)
for i in Iterable2:
    print(i)
```

```python
# Recorridos Simples (3)
for i in Iterable3:
    print(i,Iterable3[i])
```

```python
# Recorridos Compuestos (Anidados)
Iterable_Supremo = [Iterable1,Iterable2,Iterable3]

for iterable in Iterable_Supremo:
    for i in iterable:
        print(i)
    print("*"*30)
```

#### Recorridos tipo: while

```python
# Recorrido tipo: while (1)
Condicion_Logica = True
A = 1

while Condicion_Logica:
    print(A)

    A+=1
    if A>4:
        Condicion_Logica = False
```

```python
# Recorrido tipo: while (2)
A = 1
while A<10:
    print(A)
    A+=1 
```

### Uso básico de NetworkX

```python
# Creación de un grafo vacio
G = nx.Graph()
# Agrego nodos al grafo
G.add_node("Daniela")
G.add_node("A")
print(G)
```

```python
# Agrego de forma masiva nodos
G.add_nodes_from([2,3])
G.add_nodes_from(range(10,100))
G.add_nodes_from([i for i in range(100,200)])
print(G)
```

```python
# Agrego nodos con atributos
G.add_nodes_from([(4, {"color": "red"}),
(5, {"color": "green","estatura":10})])
print(G)
# Listado con contenido tupla. Tupla con estructura (node_name,node_attribute_dict)
```

```python
# Creación de un grafo vacio (Undirected)
G_UD = nx.Graph()

# Creación de enlaces (edges)
G_UD.add_edge("A","B")
G_UD.add_edge("B","C")
print(G_UD)
```

```python
# Creación de un grafo vacio (Directed)
G_D = nx.DiGraph() 

# Creación de enlaces (edges)
G_D.add_edge("A","B")
G_D.add_edge("B","C")
print(G_D)
```

```python
# Creación de un grafo vacio (Weihted)
G_DW = nx.DiGraph() 

# Creación de enlaces (edges)
G_DW.add_edge("A","B",weight = 6)
G_DW.add_edge("B","C")
print(G_DW) 
```

```python
# Creación de un grafo vacio (Signed)
G_DS = nx.DiGraph() 

# Creación de enlaces (edges)
G_DS.add_edge("A","B",sign = "+")
G_DS.add_edge("B","C",sign = "-")
print(G_DS)
```

```python
# Creación de un grafo vacio (Con otros atributos, e.g. relación)
G_UDR = nx.Graph() 

# Creación de enlaces (edges)
G_UDR.add_edge("A","B",relation = "friend")
G_UDR.add_edge("B","C",relation = "neighbor")
print(G_UDR)
```

```python
# Creación de un grafo vacio (Multigrafo:)
G_M = nx.MultiGraph() 

# Creación de enlaces (edges)
G_M.add_edge("A","B",relation = "friend")
G_M.add_edge("A","B",relation = "neighbor")
print(G_M)
```

```python
# Introducción masiva de enlaces

G = nx.Graph()
# Sin atributos
G.add_edges_from([(1, 2), (1, 3)])
# Con atributos
G.add_edges_from([(1, 4,{"weight":3,"sign":"+"}), 
(2, 3,{"weight":4,"sign":"-"})])

print(G)
```

```python
# Verificación del número de caminos que llegan de un nodo i a un nodo j, de longitud n

# Creación del grafo
G = nx.Graph()
G.add_edges_from([(1, 2), (1, 3)])

# Creación de la matriz de adyacencia
A_G = nx.to_numpy_array(G)

print('La matriz de Adyacencia es:\n\n',A_G,'\n\n'+'-'*30)

# Recorrido por el máximo número de enlaces que puede tener un camnio entre un nodo y otro antes de volverse un ciclo 
Nodos = list(G.nodes)
Lista_Matrices = []
for i in range(len(Nodos)):
    print("Caminos de longitud {}".format(i+1))
    if i == 0:
        B = A_G
    else:
        B = np.matmul(A_G,B)
    print(B)
    print("\n{}\n".format("_"*20))
    Lista_Matrices.append(B)

# Usamos entonces len(Nodos) o len(Nodos)-1?
```
